```python
import pymysql
import turtle
try:
    # 1. MySQL connection
    conn = pymysql.connect(
        host='localhost',
        user='root',       
        password='root', 
        database='student'      
    )
    cursor = conn.cursor()
    # 2. Table create 
    cursor.execute('''
    CREATE TABLE IF NOT EXISTS users (
        id INT AUTO_INCREMENT PRIMARY KEY,
        name VARCHAR(255),
        email VARCHAR(255)
    )
    ''')
    # 3. Turtle screen setup
    screen = turtle.Screen()
    # 4. Input fields
    name = screen.textinput("User Form", "Enter your name:")
    email = screen.textinput("User Form", "Enter your email:")
    # 5. Data insert 
    if name and email:
        sql = "INSERT INTO users (name, email) VALUES (%s, %s)"
        cursor.execute(sql, (name, email))
        conn.commit()
        turtle.write("Saved to MySQL!", align="center", font=("Arial", 16, "bold"))
    else:
        turtle.write("Missing input!", align="center", font=("Arial", 16, "bold"))
    # 6. Turtle window open 
    turtle.done()
except Exception as e:
    print("Error:", e)
finally:
    # 7. Database connection close 
    if 'conn' in locals():
        conn.close()
```
